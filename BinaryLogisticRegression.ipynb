```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X , y = load_breast_cancer(return_X_y=True)

X_train , X_test, y_train, y_test =train_test_split(X, y,
                                                    test_size=0.2,
                                                    random_state=1)


X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
class BinaryLogisticRegression():
    def __init__(self) -> None:
        self.__sigmoid = (lambda z : 1.0/ (1.0 + np.exp(-z)))
        self.__theta = None

    def cal_gradient(self, X, y) :
        m = y.size
        return ((X.T @ ( self.__sigmoid(X @ self.__theta) - y)) / m)

    def fit(self, X, y, lr=0.1, epochs=100 ):

        X_b = np.c_[np.ones((X.shape[0],1)), X]
        self.__theta = np.zeros(X_b.shape[1])

        for epoch in range(epochs):
            self.__theta -= lr * self.cal_gradient(X_b, y)

    def predict(self, X, theshole=0.5) :
        X_b = np.c_[np.ones((X.shape[0],1)), X]
        return (self.__sigmoid(X_b @ self.__theta) >= theshole).astype(int)

```

```python
from sklearn.metrics import accuracy_score
logistReg = BinaryLogisticRegression()
logistReg.fit(X_train, y_train, epochs=100000)
```

```python
y_pred = logistReg.predict(X_test)

print(accuracy_score(y_test, y_pred))
```
